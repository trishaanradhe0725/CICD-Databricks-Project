```python
# Sample data: (sale_id, date, product, unit_price, quantity)
data = [
    (1, "2025-01-01", "Laptop", 1200.0, 2),
    (2, "2025-01-02", "Smartphone", 800.0, 5),
    (3, "2025-01-02", "Tablet", 400.0, 3),
    (4, "2025-01-03", "Headphones", 150.0, 10),
    (5, "2025-01-04", "Laptop", 1200.0, 1),
    (6, "2025-01-05", "Smartphone", 800.0, 4),
    (7, "2025-01-06", "Monitor", 300.0, 2),
    (8, "2025-01-07", "Keyboard", 80.0, 15)
]

# Define column names
columns = ["sale_id", "date", "product", "unit_price", "quantity"]
from pyspark.sql.types import *

# Optional: Define explicit schema for better control
schema = StructType([
    StructField("sale_id", IntegerType(), True),
    StructField("date", StringType(), True),      # Could use DateType() if parsing dates
    StructField("product", StringType(), True),
    StructField("unit_price", DoubleType(), True),
    StructField("quantity", IntegerType(), True)
])

# Create DataFrame
sales_df = spark.createDataFrame(data, schema=schema)
# Or without explicit schema: sales_df = spark.createDataFrame(data, columns)

```
